```python
#correlation analysis
numeric_df = df.select_dtypes(include=[np.number])
```

```python
# Histograms for selected numeric columns
for col in ['ev_sales', 'petrol_car_sales', 'diesel_car_sales', 'total_vehicle_sales', 'ev_market_share']:
    if col in df.columns:
        plt.figure()
        sns.histplot(df[col], kde=True, bins=30)
        plt.title(f'Histogram of {col}')
        plt.xlabel(col)
        plt.ylabel('Frequency')
        plt.tight_layout()
        plt.show()
```

```python
# Pie Chart or count plot for a categorical variable 'region'
if 'region' in df.columns:
    plt.figure()
    sns.countplot(data=df, x='region')
    plt.title('Count Plot of Regions')
    plt.xlabel('Region')
    plt.ylabel('Count')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
# Box Plot to inspect distribution of EV market share
if 'ev_market_share' in df.columns:
    plt.figure()
    sns.boxplot(x=df['ev_market_share'])
    plt.title('Box Plot of EV Market Share')
    plt.tight_layout()
    plt.show()
```

```python
#Global Sales Trend (Yearly)
yearly_sales = df.groupby("year")[["ev_sales", "petrol_car_sales", "diesel_car_sales"]].sum()

plt.figure(figsize=(12,6))
yearly_sales.plot()
plt.title("EV vs Petrol vs Diesel Sales Over Time")
plt.ylabel("Sales")
plt.show()
```

```python
#EV Market Share Trend
ev_share_year = df.groupby("year")["ev_market_share"].mean()

plt.figure(figsize=(10,5))
ev_share_year.plot(marker="o", color="green")
plt.title("Average EV Market Share Over Time")
plt.ylabel("EV Market Share (%)")
plt.show()
```

```python
#Country-Level EV Dominance
dominant_countries = df[df["is_ev_dominant"] == 1]["country"].unique()
print("Countries where EV became dominant:")
print(dominant_countries)
```

```python
# correlation heatmap if we have four or more numeric columns
if numeric_df.shape[1] >= 4:
    plt.figure(figsize=(12, 10))
    correlation_matrix = numeric_df.corr()
    sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
    plt.title('Correlation Heatmap of Numeric Features')
    plt.tight_layout()
    plt.show()
```

```python
#Compare Dominant vs Non-Dominant Countries
df.groupby("is_ev_dominant")[[
    "charging_stations",
    "ev_subsidy_usd",
    "emission_regulation_score",
    "fuel_price_usd_per_liter",
    "gdp_per_capita"
]].mean()
```

```python
#EV Market Share per Region over Years
region_year_share = (
    df.groupby(['region', 'year'])['ev_market_share']
    .mean()
    .unstack()
)

plt.figure()
region_year_share.T.plot()
plt.title("EV Market Share by Region Over Years")
plt.xlabel("Year")
plt.ylabel("EV Market Share")
plt.grid()
plt.show()
```

```python
#EV Market Share by Vehicle Segment & Powertrain
segment_share = (
    df.groupby(['vehicle_segment', 'powertrain_type'])
    .agg(avg_share=('ev_market_share', 'mean'))
    .reset_index()
)

pivot_seg = segment_share.pivot(index='vehicle_segment',
                               columns='powertrain_type',
                               values='avg_share')

plt.figure()
pivot_seg.plot(kind='bar')
plt.title("EV Market Share by Segment & Powertrain")
plt.ylabel("Avg Market Share")
plt.xticks(rotation=45)
plt.show()
```

```python
#Fast Charger Share Trend
fast_charger_trend = (
    df.groupby('year')['fast_chargers_share']
    .mean()
)

plt.figure()
plt.plot(fast_charger_trend.index, fast_charger_trend.values)
plt.title("Fast Charger Share Over Time")
plt.xlabel("Year")
plt.ylabel("Fast Charger %")
plt.grid()
plt.show()
```

```python
#EV Range Improvement Over Years
ev_range = df.groupby('year')['avg_ev_range_km'].mean()

plt.figure()
plt.plot(ev_range.index, ev_range.values)
plt.title("EV Range Improvement Over Time")
plt.xlabel("Year")
plt.ylabel("Average Range (km)")
plt.grid()
plt.show()
```

```python
#CO2 vs EV Sales Trend
co2_trend = df.groupby('year').agg(
    ev_sales=('ev_sales', 'sum'),
    co2=('co2_emissions_transport_mt', 'mean')
)

plt.figure()
plt.plot(co2_trend.index, co2_trend['ev_sales'], label='EV Sales')
plt.plot(co2_trend.index, co2_trend['co2'], label='CO2 Emissions')
plt.title("EV Sales vs CO2 Emissions")
plt.xlabel("Year")
plt.legend()
plt.grid()
plt.show()
```

```python
#Growth Rate per Region Over Years
growth_region = (
    df.groupby(['region', 'year'])['ev_growth_rate_yoy']
    .mean()
    .unstack()
)

plt.figure()
growth_region.T.plot()
plt.title("EV Growth Rate by Region Over Years")
plt.xlabel("Year")
plt.ylabel("Growth Rate %")
plt.grid()
plt.show()
```

```python
#EV Dominant Countries
ev_dom = df[df['is_ev_dominant'] == 1]

counts = ev_dom['country'].value_counts()

plt.figure()
counts.plot(kind='bar')
plt.title("Countries Where EV is Dominant")
plt.xlabel("Country")
plt.ylabel("Count")
plt.show()
```

```python
#Top CO2 Emitting Countries
#long-term growth (CAGR)—this is critical for forecasting and strategy
co2_2025 = (
    df[df['year'] == 2025]
    .groupby('country')['co2_emissions_transport_mt']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure()
co2_2025.plot(kind='bar')
plt.title("Top CO2 Emitting Countries (2025)")
plt.ylabel("CO2 Emissions")
plt.show()
```

```python
#EV Adoption vs Infrastructure Efficiency
#How many EVs per charging station?
infra_eff = df.groupby('year').agg(
    total_ev_sales=('ev_sales', 'sum'),
    total_chargers=('charging_stations', 'sum')
).reset_index()

infra_eff['ev_per_charger'] = infra_eff['total_ev_sales'] / infra_eff['total_chargers']

plt.figure()
plt.plot(infra_eff['year'], infra_eff['ev_per_charger'])
plt.title("EVs per Charging Station (Infrastructure Load)")
plt.xlabel("Year")
plt.ylabel("EVs per Charger")
plt.grid()
plt.show()
```

```python
#EV vs ICE Market Share Ratio
#normalized ratio KPI
sales = df.groupby('year').agg(
    ev=('ev_sales', 'sum'),
    ice=('petrol_car_sales', 'sum')
).reset_index()

sales['ev_share_ratio'] = sales['ev'] / (sales['ev'] + sales['ice'])

plt.figure()
plt.plot(sales['year'], sales['ev_share_ratio'])
plt.title("EV Share vs ICE Over Time")
plt.xlabel("Year")
plt.ylabel("EV Share Ratio")
plt.grid()
plt.show()
```

```python
#CO₂ Reduction Efficiency
#How much CO₂ is reduced per EV adoption?
co2_eff = df.groupby('year').agg(
    co2=('co2_emissions_transport_mt', 'mean'),
    ev_share=('ev_market_share', 'mean')
).reset_index()

co2_eff['co2_per_ev_share'] = co2_eff['co2'] / co2_eff['ev_share']

plt.figure()
plt.plot(co2_eff['year'], co2_eff['co2_per_ev_share'])
plt.title("CO2 Emissions per EV Market Share Unit")
plt.xlabel("Year")
plt.ylabel("CO2 per EV Share")
plt.grid()
plt.show()
```

```python
#Market Concentration (Top Countries Dependency)
#dependency risk
country_sales = df.groupby('country')['ev_sales'].sum().sort_values(ascending=False)

top5_share = country_sales.head(5).sum() / country_sales.sum()
print(f"Top 5 Countries Share: {top5_share:.2%}")

# Plot
plt.figure()
country_sales.head(10).plot(kind='bar')
plt.title("Top 10 Countries EV Sales")
plt.xlabel("Country")
plt.ylabel("EV Sales")
plt.show()
```

```python
#Price Sensitivity Curve (Fuel Price vs EV Share)
#relationship shape
plt.figure()
plt.scatter(df['fuel_price_usd_per_liter'], df['ev_market_share'])
plt.title("Fuel Price vs EV Market Share")
plt.xlabel("Fuel Price")
plt.ylabel("EV Market Share")
plt.grid()
plt.show()
```

```python
#Lagging vs Leading Regions (Gap KPI)
#gap between best and worst
region_share = df.groupby('region')['ev_market_share'].mean()

gap = region_share.max() - region_share.min()
print(f"Adoption Gap: {gap:.2f}")

plt.figure()
region_share.plot(kind='bar')
plt.title("EV Market Share by Region")
plt.ylabel("Avg EV Share")
plt.show()
```

```python
#Technology Impact KPI (Range vs Adoption)
#Does better range increase adoption?
plt.figure()
plt.scatter(df['avg_ev_range_km'], df['ev_market_share'])
plt.title("EV Range vs Market Share")
plt.xlabel("Range (km)")
plt.ylabel("EV Market Share")
plt.grid()
plt.show()
```
