```python
import pandas as pd
import plotly.graph_objects as go

# ==========================================
# LOAD DATA
# ==========================================

df = pd.read_excel(
    r"C:\Users\ahmed3farag\Downloads\RAW DATA-ev.xlsx"
)

# ==========================================
# CLEAN COLUMN NAMES
# ==========================================

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

# ==========================================
# COLUMN NAMES
# ==========================================

gdp_col = 'gdp_per_capita'

urban_col = 'urban_population_percent'

year_col = 'year'

# ==========================================
# CONVERT TO NUMERIC
# ==========================================

for col in [gdp_col, urban_col, year_col]:

    df[col] = pd.to_numeric(
        df[col],
        errors='coerce'
    )

# ==========================================
# REMOVE NULLS
# ==========================================

df = df.dropna(
    subset=[gdp_col, urban_col, year_col]
)

# ==========================================
# GROUP BY YEAR
# ==========================================

yearly_data = (
    df.groupby(year_col)[
        [gdp_col, urban_col]
    ]
    .mean()
    .reset_index()
)

# ==========================================
# CREATE CHART
# ==========================================

fig = go.Figure()

# ------------------------------------------
# URBAN POPULATION BAR (BACKGROUND)
# ------------------------------------------

fig.add_trace(

    go.Bar(

        x=yearly_data[year_col],

        y=yearly_data[urban_col],

        name='Urban Population %',

        marker_color='#2d8bba',

        opacity=0.35,

        yaxis='y2'
    )
)

# ------------------------------------------
# GDP LINE (FRONT)
# ------------------------------------------

fig.add_trace(

    go.Scatter(

        x=yearly_data[year_col],

        y=yearly_data[gdp_col],

        name='GDP per Capita',

        mode='lines+markers',

        line=dict(
            color='#00b58f',
            width=4
        ),

        marker=dict(
            size=8,
            color='#00b58f'
        ),

        yaxis='y1'
    )
)

# ==========================================
# LAYOUT
# ==========================================

fig.update_layout(

    title=' Economic & Urbanization Trends',

    template='plotly_white',

    xaxis=dict(
        title='Year'
    ),

    yaxis=dict(
        title='GDP per Capita ($)'
    ),

    yaxis2=dict(

        title='Urban Population %',

        overlaying='y',

        side='right'
    ),

    legend=dict(
        x=0.01,
        y=0.99
    ),

    height=600
)

# ==========================================
# SHOW OUTPUT
# ==========================================

fig.show()
```

```python
import plotly.graph_objects as go
# ==========================================
# COLUMN NAMES
# ==========================================

year_col = 'year'

stations_col = 'charging_stations'

fast_col = 'fast_chargers_share'

# ==========================================
# CONVERT TO NUMERIC
# ==========================================

for col in [year_col, stations_col, fast_col]:

    df[col] = pd.to_numeric(
        df[col],
        errors='coerce'
    )
# ==========================================
# GROUP BY YEAR
# ==========================================

infra_data = (
    df.groupby(year_col)[
        [stations_col, fast_col]
    ]
    .mean()
    .reset_index()
)

# ==========================================
# CREATE PROFESSIONAL CHART
# ==========================================

fig = go.Figure()

# ------------------------------------------
# CHARGING STATIONS BAR
# ------------------------------------------

fig.add_trace(

    go.Bar(

        x=infra_data[year_col],

        y=infra_data[stations_col],

        name='Charging Stations',

        marker_color='#6c5ce7',

        opacity=0.75,

        yaxis='y1'
    )
)

# ------------------------------------------
# FAST CHARGERS SHARE LINE
# ------------------------------------------

fig.add_trace(

    go.Scatter(

        x=infra_data[year_col],

        y=infra_data[fast_col],

        name='Fast Chargers Share %',

        mode='lines+markers',

        line=dict(
            color='#00cec9',
            width=4
        ),

        marker=dict(
            size=8,
            color='#00cec9'
        ),

        yaxis='y2'
    )
)

# ==========================================
# LAYOUT
# ==========================================

fig.update_layout(

    title=' EV Infrastructure & Fast Charging Trends',

    template='plotly_white',

    xaxis=dict(
        title='Year'
    ),

    yaxis=dict(
        title='Charging Stations'
    ),

    yaxis2=dict(

        title='Fast Chargers Share %',

        overlaying='y',

        side='right'
    ),

    legend=dict(
        x=0.01,
        y=0.99
    ),

    height=600
)

# ==========================================
# SHOW OUTPUT
# ==========================================

fig.show()
```

```python
import plotly.graph_objects as go

# ==========================================
# KPI VALUES
# ==========================================

fuel_price = 1.342239
electricity_price = 0.1695078

# ==========================================
# CREATE FIGURE
# ==========================================

fig = go.Figure()

# ==========================================
# FUEL PRICE KPI
# ==========================================

fig.add_trace(go.Indicator(

    mode="gauge+number",

    value=fuel_price,

    title={
        'text': "⛽ Average Fuel Price (USD/L)"
    },

    domain={
        'x': [0, 0.45],
        'y': [0, 1]
    },

    number={
        'valueformat': ".3f",
        'suffix': " $"
    },

    gauge={

        'axis': {
            'range': [0, 2]
        },

        'bar': {
            'thickness': 0.3,
            'color': "#e74c3c"
        },

        'steps': [

            {
                'range': [0, 0.7],
                'color': "#f8d7da"
            },

            {
                'range': [0.7, 1.4],
                'color': "#f5b7b1"
            },

            {
                'range': [1.4, 2],
                'color': "#c0392b"
            }
        ]
    }
))

# ==========================================
# ELECTRICITY PRICE KPI
# ==========================================

fig.add_trace(go.Indicator(

    mode="gauge+number",

    value=electricity_price,

    title={
        'text': "⚡ Average Electricity Price (USD/kWh)"
    },

    domain={
        'x': [0.55, 1],
        'y': [0, 1]
    },

    number={
        'valueformat': ".3f",
        'suffix': " $"
    },

    gauge={

        'axis': {
            'range': [0, 1]
        },

        'bar': {
            'thickness': 0.3,
            'color': "#2ecc71"
        },

        'steps': [

            {
                'range': [0, 0.3],
                'color': "#d4efdf"
            },

            {
                'range': [0.3, 0.6],
                'color': "#abebc6"
            },

            {
                'range': [0.6, 1],
                'color': "#27ae60"
            }
        ]
    }
))

# ==========================================
# LAYOUT
# ==========================================

fig.update_layout(

    template='plotly_white',

    height=500,

    width=1100,

    margin=dict(
        t=100,
        b=30,
        l=30,
        r=30
    ),

    title={

        'text': '🚗 EV vs Petrol Energy Cost Analysis',

        'x': 0.5,

        'font': {
            'size': 26
        }
    },

    font={
        'family': 'Arial'
    }
)

# ==========================================
# SHOW FIGURE
# ==========================================

fig.show()
```
